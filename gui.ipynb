```python
%%time

import tkinter as tk

def add_numbers():
    number1 = float(entry1.get())
    number2 = float(entry2.get())

    answer = number1 + number2

    result.config(text="Answer: " + str(answer))


window = tk.Tk()
window.title("Addition Calculator")
window.geometry("300x250")

label1 = tk.Label(window, text="Enter first number:")
label1.pack()

entry1 = tk.Entry(window)
entry1.pack()

label2 = tk.Label(window, text="Enter second number:")
label2.pack()

entry2 = tk.Entry(window)
entry2.pack()

add_button = tk.Button(window, text="Add", command=add_numbers)
add_button.pack()

result = tk.Label(window, text="Answer:")
result.pack()

window.mainloop()
```
